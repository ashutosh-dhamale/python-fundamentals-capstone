```python

# Importing pandas library
import pandas as pd

# Importing numpy library
import numpy as np
```

```python

# Creating Employee DataFrame
employee = pd.DataFrame({

    # Creating ID column
    'ID': ['A001', 'A002', 'A003', 'A004', 'A005'],

    # Creating Name column
    'Name': ['John Alter', 'Alice Luxumberg', 'Tom Sabestine', 'Nina Adgra', 'Amy Johny'],

     # Creating Gender column
    'Gender': ['M', 'F', 'M', 'F', 'F'],

     # Creating Gender column
    'City': ['Paris', 'London', 'Berlin', 'Newyork', 'Madrid'],

    # Creating Age column
    'Age': [25, 27, 29, 31, 30]
})

# Displaying employee dataframe
employee
```

```python

# Creating Seniority Level DataFrame
seniority = pd.DataFrame({

    # Creating ID column
    'ID': ['A001', 'A002', 'A003', 'A004', 'A005'],

    # Creating Designation Level column
    'Designation Level': [2, 2, 3, 2, 3]
})

# Displaying seniority dataframe
seniority
```

```python

# Creating Project DataFrame
project = pd.DataFrame({

     # Creating ID column
    'ID': ['A001','A002','A003','A004','A005','A002','A005','A003','A001','A003','A001','A004','A004','A005'],

     # Creating Project column
     'Project': ['Project 1','Project 2','Project 3','Project 4','Project 5','Project 6','Project 7','Project 8','Project 9','Project 10','Project 11','Project 12','Project 13','Project 14'],

      # Creating Cost column with missing values
     'Cost': [1002000,2000000,4500000,5500000,np.nan,680000,400000,350000,np.nan,300000,2000000,1000000,3000000,200000],

     # Creating Status column
     'Status': ['Finished','Ongoing','Finished','Ongoing','Finished','Failed','Finished','Failed','Ongoing','Finished','Failed','Ongoing','Finished','Finished']
})

# Displaying project dataframe
project
```

```python

# Saving employee dataframe as CSV file
employee.to_csv('employee.csv', index=False)

# Saving seniority dataframe as CSV file
seniority.to_csv('seniority.csv', index=False)

# Saving project dataframe as CSV file
project.to_csv('project.csv', index=False)
```

```python

# Reading project CSV file
proj = pd.read_csv('project.csv')

# Initializing total variable
total = 0

# Initializing count variable
count = 0

# Running loop through project dataframe
for i in range(len(proj)):

    # Checking if cost value is not missing
    if not pd.isna(proj.loc[i, 'Cost']):

        # Adding existing cost to total
        total += proj.loc[i, 'Cost']

        # Increasing count
        count += 1

     # Executing if value is missing
    else:

        # Replacing missing value with running average
        proj.loc[i, 'Cost'] = total / count

# Displaying updated dataframe
proj
```

```python


# Saving updated project dataframe
proj.to_csv('project.csv', index=False)

```

```python

# Reading employee CSV file
emp = pd.read_csv('employee.csv')

# Splitting Name column into First Name and Last Name
emp[['First Name', 'Last Name']] = emp['Name'].str.split(' ', expand=True)

# Dropping original Name colum
emp.drop('Name', axis=1, inplace=True)

# Displaying employee dataframe
emp
```

```python

# Reading seniority CSV file
sen = pd.read_csv('seniority.csv')


# Reading updated project CSV file
proj = pd.read_csv('project.csv')

# Merging employee dataframe and seniority dataframe
df1 = pd.merge(emp, sen, on='ID')

# Merging project dataframe with previous merged dataframe
final = pd.merge(df1, proj, on='ID')

# Display final dataframe
final
```

```python

# Adding Bonus column for finished projects
final['Bonus'] = np.where(final['Status'] == 'Finished',
                         final['Cost'] * 0.05, 0)

# Displaying final dataframe
final
```

```python

# Increase designation level for failed projects
final.loc[final['Status'] == 'Failed', 'Designation Level'] += 1

# Remove employees with designation level > 4
final = final[final['Designation Level'] <= 4]

# Displaying updated dataframe
final
```

```python

# Defining function to add title
def add_title(row):

      # Checking male gender
    if row['Gender'] == 'M':

        # Adding Mr. before first name
        return 'Mr. ' + row['First Name']
    else:

        # Adding Mrs. before first name
        return 'Mrs. ' + row['First Name']

# Applying function on First Name column
final['First Name'] = final.apply(add_title, axis=1)

# Dropping Gender column
final.drop('Gender', axis=1, inplace=True)

# Displaying updated dataframe
final
```

```python


# Promoting employees whose age is greater than 29
final.loc[final['Age'] > 29, 'Designation Level'] -= 1

# Ensuring designation level does not go below 1
final.loc[final['Designation Level'] < 1, 'Designation Level'] = 1

# Displaying updated dataframe
final

```

```python

# Grouping dataframe by ID and First Name
total_cost = final.groupby(['ID', 'First Name'])['Cost'].sum().reset_index()

# Renaming Cost column to Total Cost
total_cost.rename(columns={'Cost': 'Total Cost'}, inplace=True)

# Displaying total project cost dataframe
total_cost
```

```python

# Filtering employees whose city contains letter o
result = final[final['City'].str.contains('o', case=False)]

# Displaying filtered dataframe
result
```

```python

# Saving final dataframe as CSV file
final.to_csv('final_output.csv', index=False)

# Saving total project cost dataframe as CSV file
total_cost.to_csv('total_project_cost.csv', index=False)
```
